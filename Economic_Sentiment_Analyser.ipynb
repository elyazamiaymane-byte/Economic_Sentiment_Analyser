```python
# ============================================
# Economic Sentiment Analyser
# By El Yazami Mohamed Aymane
# ============================================

from collections import Counter
import re

# Economic sentiment lexicons
POSITIVE_WORDS = {
    "growth", "surplus", "recovery", "expansion", "profit", "gain",
    "increase", "rise", "improvement", "strong", "stable", "boost",
    "opportunity", "investment", "innovation", "productivity", "demand",
    "confidence", "optimism", "development", "reform", "progress"
}

NEGATIVE_WORDS = {
    "recession", "deficit", "decline", "loss", "decrease", "fall",
    "crisis", "debt", "inflation", "unemployment", "risk", "weak",
    "contraction", "slowdown", "default", "instability", "poverty",
    "shortage", "downturn", "collapse", "stagnation", "austerity"
}

ECONOMIC_SIGNALS = {
    "gdp", "inflation", "interest", "rate", "policy", "market",
    "trade", "export", "import", "currency", "fiscal", "monetary",
    "budget", "revenue", "tax", "bond", "equity", "reserve"
}

def clean_and_tokenize(text):
    text = text.lower()
    tokens = re.findall(r'\b[a-z]+\b', text)
    stopwords = {"the", "a", "an", "and", "or", "but", "in", "on",
                 "at", "to", "for", "of", "with", "is", "are", "was",
                 "were", "has", "have", "had", "be", "been", "that",
                 "this", "it", "its", "by", "from", "as", "we", "our"}
    return [t for t in tokens if t not in stopwords]

def analyse_sentiment(tokens):
    pos = [t for t in tokens if t in POSITIVE_WORDS]
    neg = [t for t in tokens if t in NEGATIVE_WORDS]
    signals = [t for t in tokens if t in ECONOMIC_SIGNALS]

    score = len(pos) - len(neg)
    if score > 0:
        label = "POSITIVE 📈"
    elif score < 0:
        label = "NEGATIVE 📉"
    else:
        label = "NEUTRAL ➡️"

    return pos, neg, signals, score, label

def display_bar(word, count, max_count, width=30):
    bar_len = int((count / max_count) * width)
    bar = "█" * bar_len
    return f"  {word:<20} {bar} {count}"

def analyse(text):
    print("=" * 55)
    print("       ECONOMIC TEXT SENTIMENT ANALYSER")
    print("=" * 55)

    tokens = clean_and_tokenize(text)
    if not tokens:
        print("No meaningful tokens found.")
        return

    freq = Counter(tokens)
    pos, neg, signals, score, label = analyse_sentiment(tokens)

    # Word frequency
    print(f"\n📊 WORD FREQUENCY (top 10)")
    print("-" * 55)
    top10 = freq.most_common(10)
    max_count = top10[0][1]
    for word, count in top10:
        print(display_bar(word, count, max_count))

    # Sentiment
    print(f"\n🧠 SENTIMENT ANALYSIS")
    print("-" * 55)
    print(f"  Positive signals : {', '.join(pos) if pos else 'none'}")
    print(f"  Negative signals : {', '.join(neg) if neg else 'none'}")
    print(f"  Sentiment score  : {score:+d}")
    print(f"  Overall verdict  : {label}")

    # Economic signals
    print(f"\n💹 ECONOMIC KEYWORDS DETECTED")
    print("-" * 55)
    print(f"  {', '.join(set(signals)) if signals else 'none detected'}")

    # Summary
    print(f"\n📋 SUMMARY")
    print("-" * 55)
    print(f"  Total tokens     : {len(tokens)}")
    print(f"  Unique words     : {len(freq)}")
    print(f"  Positive terms   : {len(pos)}")
    print(f"  Negative terms   : {len(neg)}")
    print("=" * 55)

# ── Test with a sample economic text ──────────────────
sample = """
The central bank announced a strong recovery in GDP growth
following months of economic expansion. Exports increased
significantly, boosting confidence in the market. However,
inflation remains a risk and fiscal deficit continues to
weigh on investor sentiment. The government confirmed new
investment in productivity and reform to address unemployment.
"""

analyse(sample)
```

```python
# ── Interactive mode ──
text = input("Paste your economic text here: ")
analyse(text)
```

```python
# ── Comparison Mode ──────────────────────────────────
def compare(text1, text2, label1="Text A", label2="Text B"):
    tokens1 = clean_and_tokenize(text1)
    tokens2 = clean_and_tokenize(text2)

    _, _, _, score1, label_1 = analyse_sentiment(tokens1)
    _, _, _, score2, label_2 = analyse_sentiment(tokens2)

    freq1 = Counter(tokens1)
    freq2 = Counter(tokens2)

    print("=" * 55)
    print("         COMPARATIVE ECONOMIC ANALYSIS")
    print("=" * 55)

    print(f"\n  {label1:<25} vs  {label2}")
    print("-" * 55)
    print(f"  Sentiment score  : {score1:>+3}              {score2:>+3}")
    print(f"  Verdict          : {label_1:<16}   {label_2}")
    print(f"  Total tokens     : {len(tokens1):<16} {len(tokens2)}")
    print(f"  Unique words     : {len(freq1):<16} {len(freq2)}")

    # Shared keywords
    shared = set(tokens1) & set(tokens2)
    print(f"\n🔗 SHARED KEYWORDS")
    print("-" * 55)
    print(f"  {', '.join(shared) if shared else 'none'}")

    # Unique to each
    only1 = set(tokens1) - set(tokens2)
    only2 = set(tokens2) - set(tokens1)
    print(f"\n🔵 UNIQUE TO {label1.upper()}")
    print("-" * 55)
    print(f"  {', '.join(list(only1)[:10])}")
    print(f"\n🔴 UNIQUE TO {label2.upper()}")
    print("-" * 55)
    print(f"  {', '.join(list(only2)[:10])}")
    print("=" * 55)

# ── Run comparison ────────────────────────────────────
fed_statement = """
The labor market remains solid and inflation has moved closer
to the Committee's 2 percent objective. Economic uncertainty
has increased however and the risks of higher unemployment
and higher inflation have risen. The Committee decided to
maintain the target range for the federal funds rate.
"""

hormuz_crisis = """
The closure of the Strait of Hormuz threatens global oil supply
and energy markets. Tanker traffic has collapsed causing severe
disruption to export routes. Oil prices have surged creating
inflationary pressure. Fiscal deficits are expected to widen.
Investors face significant risk and market confidence has
deteriorated sharply. The economic contraction could trigger
recession in vulnerable nations.
"""

compare(fed_statement, hormuz_crisis, "Fed Statement", "Hormuz Crisis")
```
